```python
import os
import shutil
import random

base = r"D:\TEAM IDEATORS\1_objective\animal_dataset"
classes = ["elephant", "cow", "dog"]
split_ratio = 0.8  # 80% train, 20% val

for cls in classes:
    img_src = os.path.join(base, "images", cls)
    lbl_src = os.path.join(base, "labels", cls)

    images = [f for f in os.listdir(img_src) if f.endswith(".jpg")]
    random.shuffle(images)

    split = int(len(images) * split_ratio)
    train_imgs = images[:split]
    val_imgs = images[split:]

    for mode, files in [("train", train_imgs), ("val", val_imgs)]:
        img_dst = os.path.join(base, "images", mode, cls)
        lbl_dst = os.path.join(base, "labels", mode, cls)

        os.makedirs(img_dst, exist_ok=True)
        os.makedirs(lbl_dst, exist_ok=True)

        for img in files:
            shutil.move(
                os.path.join(img_src, img),
                os.path.join(img_dst, img)
            )
            shutil.move(
                os.path.join(lbl_src, img.replace(".jpg", ".txt")),
                os.path.join(lbl_dst, img.replace(".jpg", ".txt"))
            )

print("✅ Train–Validation split completed")
```
