```python
import cv2
import os

video_path = r"D:\TEAM IDEATORS\1_objective\cow\ex\🇱🇰 Cows in Kurunegala #srilanka #wildlife #cow #cows #kurunegala.mp4"
output_folder = r"D:\TEAM IDEATORS\1_objective\cow\ex"

os.makedirs(output_folder, exist_ok=True)

cap = cv2.VideoCapture(video_path)

total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
target_images = 70

step = max(1, total_frames // target_images)

frame_id = 0
saved = 0

while cap.isOpened() and saved < target_images:
    ret, frame = cap.read()
    if not ret:
        break

    if frame_id % step == 0:
        cv2.imwrite(f"{output_folder}\\cow_{saved:03d}.jpg", frame)
        saved += 1

    frame_id += 1

cap.release()
print("Total frames:", total_frames)
print("Images saved:", saved)
```
